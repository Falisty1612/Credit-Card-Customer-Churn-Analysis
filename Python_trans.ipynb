```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sakshigoyal7/credit-card-customers")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(os.path.join(path, "BankChurners.csv"))
```

```python
df.info()
df.describe()
```

```python
col_drop = [x for x in df.columns if x.startswith("Naive_Bayes")]
df = df.drop(columns=col_drop)
```

```python
df.shape
df.head()
```

```python
display(df.isna().sum())
print("Duplikaty: ",df.duplicated().sum())
```
